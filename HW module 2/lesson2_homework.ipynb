## 1. Короткий опис завдання

Потрібно вести облік домашніх витрат за категоріями, щоб бачити, скільки коштів витрачається на різні потреби. Програма має визначати загальну кількість покупок, частку витрат на окрему категорію в загальній сумі, середню вартість покупки та класифікувати покупки за розміром суми

## 2. Дані
Список щонайменше з 8 записів-словників.

```python
records = [
    {"id": 1, "name": "Продукти в АТБ", "category": "Продукти", "amount": 450},
    {"id": 2, "name": "Проїзд", "category": "Транспорт", "amount": 50},
    {"id": 3, "name": "Шампупь", "category": "Особиста гігієна", "amount": 180},
    {"id": 4, "name": "Продукти в Сільпо", "category": "Продукти", "amount": 650},
    {"id": 5, "name": "Таксі", "category": "Транспорт", "amount": 200},
    {"id": 6, "name": "Пральний порошок", "category": "Побутова хімія", "amount": 250},
    {"id": 7, "name": "Кава", "category": "Розваги", "amount": 120},
    {"id": 8, "name": "Продукти в АТБ", "category": "Продукти", "amount": 350},
    {"id": 9, "name": "Квиток в кіно", "category": "Розваги", "amount": 300},
    {"id": 10, "name": "Проїзд", "category": "Транспорт", "amount": 100}
]
print(len(records))
```

## 3. Три показники
Спершу запишіть, що очікуєте (ручний розрахунок).

1. Кількість витрат: 10
2. Частка витрат на транспорт: 350/2650 * 100 = 13,21%
3. Середня сума однієї витрати 265
4. Загальна сума витрат 2650

## 4. Класифікація (if / elif / else)
Які межі й чому?

if — якщо сума витрати до 150 грн включно, класифікувати як "Невелика"  
elif — якщо сума витрати від 151 до 300 грн включно, класифікувати як "Середня"  
else — якщо сума витрати перевищує 300 грн, класифікувати як "Велика"

## 5. Функції (щонайменше дві, з return)

calculate_average() — функція для розрахунку середньої суми однієї витрати. Отримує список витрат, обчислює загальну суму та ділить її на кількість записів. Повертає (return) середнє значення
calculate_share() — функція для розрахунку частки витрат на певну категорію. Отримує список витрат і назву категорії, обчислює суму витрат у цій категорії та її відсоток від загальної суми. Повертає (return) отриманий відсоток

```python
def calculate_average(items):
    total = sum(item["amount"] for item in items)
    return total / len(items)

def calculate_share(items, category):
    category_total = sum(item["amount"] for item in items
                         if item["category"] == category
    )
    total = sum(item["amount"]for item in items)
    return category_total / total * 100
```

## 6. Перевірки (щонайменше три assert)

1. Перевірити, що кількість записів у списку дорівнює 10.
2. Перевірити, що частка витрат на транспорт становить 13,21%.
3. Перевірити, що середня сума однієї витрати дорівнює 265 грн.

```python
print("Кількість записів:", len(records))
print("Середня сума витрат:", calculate_average(records))
print("Частка витрат на транспорт:", round(calculate_share(records, "Транспорт"), 2), "%")

for item in records:
    if item["amount"] <= 150:
        item["label"] = "Невелика"
    elif item["amount"] <= 300:
        item["label"] = "Середня"
    else:
        item["label"] = "Велика"

    print(item["name"], item["label"])

assert len(records) == 10
assert round(calculate_share(records, "Транспорт"), 2) == 13.21
assert calculate_average(records) == 265
```
